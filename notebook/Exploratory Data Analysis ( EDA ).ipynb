# Exploratory Data Analysis (EDA)

This notebook performs Univariate, Bivariate, and Multivariate analysis using Seaborn.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
df =  pd.read_csv('../data/Flight Delay and Cancellation Dataset.csv')
# Use a sample for faster plotting
sample_df = df.sample(50000, random_state=42)
```

## Univariate Analysis

### Histogram with KDE

```python
sns.histplot(sample_df['ARR_DELAY'], kde=True)
plt.show()
```

### Boxplot

```python
sns.boxplot(x=sample_df['ARR_DELAY'])
plt.show()
```

### Countplot

```python
sns.countplot(x='AIRLINE_CODE', data=sample_df)
plt.xticks(rotation=90)
plt.show()
```

## Bivariate Analysis

### Scatterplot (Numerical vs Numerical)

```python
sns.scatterplot(x='DEP_DELAY', y='ARR_DELAY', data=sample_df)
plt.show()
```

### Lineplot (Numerical vs Numerical)

```python
sns.lineplot(x='DISTANCE', y='AIR_TIME', data=sample_df)
plt.show()
```

### Boxplot (Categorical vs Numerical)

```python
sns.boxplot(x='AIRLINE_CODE', y='ARR_DELAY', data=sample_df)
plt.xticks(rotation=90)
plt.show()
```

### Barplot (Categorical vs Numerical)

```python
sns.barplot(x='AIRLINE_CODE', y='ARR_DELAY', data=sample_df)
plt.xticks(rotation=90)
plt.show()
```

## Multivariate Analysis

### Heatmap (Numerical Columns Only)

```python
numeric_df = sample_df.select_dtypes(include=['int64','float64'])

plt.figure(figsize=(12,8))
sns.heatmap(numeric_df.corr(), cmap='coolwarm')
plt.show()
```

### Scatterplot with Hue

```python
sns.scatterplot(x='DEP_DELAY', y='ARR_DELAY', hue='AIRLINE_CODE', data=sample_df)
plt.show()
```

### Lineplot with Hue

```python
sns.lineplot(x='DISTANCE', y='AIR_TIME', hue='AIRLINE_CODE', data=sample_df)
plt.show()
```
